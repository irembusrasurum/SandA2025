```python
import torch
from diffusers import StableDiffusionPipeline
import os

# 1. Modelin Yüklenmesi
# Model ID'yi aynı bırakıyoruz.
model_id = "runwayml/stable-diffusion-v1-5"

# Modelin tam hassasiyette (float32) yüklenmesini zorluyoruz, bu CPU için idealdir.
# torch_dtype=torch.float16'yı kaldırıyoruz.
pipe = StableDiffusionPipeline.from_pretrained(model_id)

# Çalışma aygıtını 'cpu' olarak belirliyoruz.
# DİKKAT: Bu, GPU'da çalıştırmaktan ÇOK daha yavaş olacaktır.
pipe = pipe.to("cpu")

# 2. Çıktı Klasörünü Oluşturma (Aynı Kalır)
output_dir = "outputs"
os.makedirs(output_dir, exist_ok=True)
print(f"Çıktılar {output_dir} klasörüne kaydedilecektir.")

# 3. İstem (Prompt) ve Görüntü Sayısı (Aynı Kalır)
prompts = [
    "A futuristic city in watercolor style, detailed, concept art",
    "An astronaut riding a horse on Mars, cinematic lighting, photo realistic",
    "A medieval library with glowing books, fantasy art, volumetric light"
]

# 4. Görüntüleri Oluşturma ve Kaydetme (Aynı Kalır)
print("Görüntüler oluşturuluyor... Lütfen bekleyin, CPU üzerinde uzun sürebilir (5-15 dakika).")
for i, prompt in enumerate(prompts):
    # Görüntüyü oluştur
    image = pipe(prompt).images[0]

    # Görüntüyü kaydet
    filename = f"{output_dir}/image_{i+1}.png"
    image.save(filename)

    print(f"'{prompt[:30]}...' istemiyle oluşturulan görüntü {filename} olarak kaydedildi.")
```
