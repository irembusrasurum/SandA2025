```python
%%writefile app.py
import streamlit as st
from transformers import pipeline

# Hafta 5'te kullandığınız modelin yüklenmesi.
# Bu dekoratör, modelin her etkileşimde yeniden yüklenmesini engeller (Hız için kritik!)
@st.cache_resource
def load_model():
    # Modelinizi ve pipeline tipini (T5 için 'text2text-generation') belirtin
    model_name = "google/flan-t5-base"

    # Modelin yüklenmesi biraz zaman alabilir
    with st.spinner('Hugging Face model yükleniyor, lütfen bekleyin...'):
        # Cihazı otomatik olarak seçer (GPU varsa GPU, yoksa CPU)
        model_pipe = pipeline(
            "text2text-generation",
            model=model_name,
            device=-1 # CPU kullanmak için -1, GPU için 0
        )
    return model_pipe

# -----------------
# Streamlit Arayüzü
# -----------------

st.set_page_config(page_title="Hafta 6 Chatbot Arayüzü", layout="centered")
st.title("🤖 Ücretsiz Model ile Streamlit Chatbot")
st.subheader("Hugging Face T5 Modeli")

# Modeli yükle
chatbot_pipeline = load_model()

# Sohbet Fonksiyonu (Hafta 5'teki optimizasyonları korur)
def get_response(prompt):
    # Repetition Penalty ve Beam Search gibi ayarlarınızı buraya ekleyin
    response = chatbot_pipeline(
        prompt,
        max_new_tokens=150,
        do_sample=False,
        num_beams=3,
        repetition_penalty=1.5
    )[0]['generated_text']
    return response

# Kullanıcı Giriş Formu
with st.form("chatbot_form"):
    # Metin giriş alanı
    user_input = st.text_input("Sorunuzu buraya yazın:", placeholder="Kuantum fiziği nedir?", key="input_text")

    # Gönderme butonu
    submitted = st.form_submit_button("Yanıt Al 🚀")

    # Yanıt Alanı
    if submitted and user_input:
        with st.spinner('Model yanıt oluşturuyor...'):
            try:
                # Modeli çağır
                response_text = get_response(user_input)

                # Yanıtı göster
                st.success("Cevap:")
                st.info(response_text)

            except Exception as e:
                st.error(f"Yanıt oluşturulurken bir hata oluştu: {e}")

st.markdown("---")
st.caption("Geliştirici: [Adınız/Kullanıcı Adınız]")
```
