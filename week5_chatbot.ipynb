```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, pipeline

# Modeli ve tokenizer'ı yükle
model_name = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Veya kolay bir kullanım için 'pipeline' kullanın
chatbot_pipeline = pipeline("text2text-generation", model=model, tokenizer=tokenizer)
```

```python
def get_chatbot_response_t5(prompt):
    response = chatbot_pipeline(
        prompt,
        max_new_tokens=150,
        do_sample=False, # Rastgeleliği kapatın (daha deterministik/mantıklı çıkarım)
        num_beams= 3, # Beam search kullanın (daha kaliteli çıktı arar)
        repetition_penalty=1.5 # 1.0'dan büyük bir değer tekrarı azaltır
    )[0]['generated_text']
    return response
```

```python
# Fonksiyon tanımı bu hücrede veya önceki bir hücrede olmalı

user_question = "What is quantum physics? Explain shortly."

# Fonksiyonu çağırın ve sonucu bir değişkene atayın
chatbot_answer = get_chatbot_response_t5(user_question)

# Sonucu ekrana yazdırın
print(f"Kullanıcı: {user_question}")
print(f"Chatbot: {chatbot_answer}")
```
