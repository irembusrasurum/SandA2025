```python
%%writefile app.py
import sys
!{sys.executable} -m pip install streamlit

import streamlit as st
from transformers import pipeline
from diffusers import StableDiffusionPipeline
import torch

# Uygulama Başlığı
st.title("AI Sanat & Sohbet Uygulaması")

# Mod Seçimi (Kenar Çubuğunda)
secilen_mod = st.sidebar.selectbox(
    "Lütfen Bir Mod Seçin:",
    ("Sohbet Modu", "Sanat Modu")
)

# Uygulamanın en üstüne, model yükleme kodundan önce ekleyin
if "messages" not in st.session_state:
    st.session_state.messages = []

@st.cache_resource
def model_yukle():
    model = pipeline("text2text-generation", model="google/flan-t5-base")
    return model

# Modeli yükle
chatbot = model_yukle()

@st.cache_resource
def gorsel_model_yukle():
    pipeline = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5",
        torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32
    )
    pipeline.to("cuda" if torch.cuda.is_available() else "cpu")
    return pipeline

# Görsel modelini yükle
gorsel_uretici = gorsel_model_yukle()

# Seçilen Moda Göre İçeriği Görüntüleme
if secilen_mod == "Sohbet Modu":
    st.header("💬 Sohbet Botu")

    # Önceki mesajları göster
    for message in st.session_state.messages:
        st.chat_message(message["role"]).write(message["content"])

    # Yeni kullanıcı girdisi
    kullanici_girisi = st.chat_input("Sorunuzu Buraya Yazın:")

    if kullanici_girisi:
        # 1. Kullanıcı mesajını geçmişe ekle ve göster
        st.session_state.messages.append({"role": "user", "content": kullanici_girisi})
        st.chat_message("user").write(kullanici_girisi)

        with st.spinner('Yanıt Üretiliyor...'):
            # 2. Bot yanıtını al
            yanit = chatbot(kullanici_girisi, max_length=50)[0]['generated_text']

            # 3. Bot yanıtını geçmişe ekle ve göster
            st.session_state.messages.append({"role": "assistant", "content": yanit})
            st.chat_message("assistant").write(yanit)

elif secilen_mod == "Sanat Modu":
    st.header("🎨 Görsel Üretici")

    sanat_istemi = st.text_area(
        "Görsel Açıklamasını Girin:",
        "A futuristic city in watercolor style, detailed, vibrant"
    )

    # Görseli üretmek için bir düğme
    if st.button("Görsel Üret"):
        if sanat_istemi:
            with st.spinner('Görsel Üretiliyor... (Bu biraz zaman alabilir)'):
                # Görsel üretme işlemi
                gorsel = gorsel_uretici(sanat_istemi).images[0]

                # Üretilen görseli Streamlit'te göster
                st.image(gorsel, caption=sanat_istemi)

                st.success("Görsel Başarıyla Üretildi!")
        else:
            st.warning("Lütfen bir görsel açıklaması girin.")
```
