### 1. Docker Temel Kavramları ve Mimarisi

**Docker Nedir?**
Docker, uygulamaları ve bağımlılıklarını izole edilmiş 'konteynerler' içinde paketleyerek hızlı ve güvenilir bir şekilde çalıştırmayı sağlayan bir açık kaynak platformudur. Bu konteynerler, uygulamaların farklı ortamlarda (geliştirme, test, üretim) tutarlı bir şekilde çalışmasını garanti eder.

**Neden Docker Kullanılır?**
*   **Tutarlılık:** Geliştirme ortamınızda çalışan bir uygulamanın üretimde de aynı şekilde çalışmasını sağlar.
*   **İzolasyon:** Uygulamalar ve bağımlılıkları birbirinden izole edilir, bu da çakışmaları önler.
*   **Taşınabilirlik:** Konteynerler, herhangi bir Docker yüklü makinede sorunsuz bir şekilde çalışabilir.
*   **Verimlilik:** Sanal makinelere göre daha hafif ve hızlıdır, kaynakları daha verimli kullanır.

**Konteynerler ve Sanal Makineler Arasındaki Farklar:**
| Özellik         | Konteynerler (Docker)                                      | Sanal Makineler (VM)                                        |
| :-------------- | :--------------------------------------------------------- | :---------------------------------------------------------- |
| **İzolasyon**   | İşletim sistemi seviyesinde izolasyon (OS'u paylaşır)     | Donanım seviyesinde tam izolasyon (kendi OS'u vardır)      |
| **Kaynak Kullanımı**| Daha hafif, OS kaynaklarını paylaşır                      | Daha ağır, her VM kendi OS'i için kaynak kullanır           |
| **Başlangıç Süresi**| Saniyeler içinde başlar                                    | Dakikalar sürebilir                                          |
| **Boyut**       | MB seviyesinde (uygulama ve bağımlılıklar)                 | GB seviyesinde (OS + uygulama + bağımlılıklar)               |
| **Temel**       | Host işletim sistemi üzerinde Docker Engine çalışır         | Hypervisor üzerinde ayrı işletim sistemleri çalışır         |

**Temel Docker Mimarisi:**
Docker, bir **istemci-sunucu (client-server)** mimarisi kullanır. Bu mimari aşağıdaki ana bileşenlerden oluşur:

*   **Docker Engine:** Docker Engine, Docker'ın temelini oluşturan, Docker imajlarını (images) oluşturan ve konteynerleri çalıştıran bileşendir. Bir **Daemon (Docker Daemon)** olarak çalışır ve API isteklerini dinler.
*   **Docker İstemcisi (Client):** Docker komutlarını (örneğin `docker run`) girmenizi sağlayan birincil yoldur. İstemci, bu komutları Docker Daemon'a REST API aracılığıyla iletir.
*   **Docker Kayıt Defteri (Registry):** Docker imajlarını depolayan ve dağıtan bir hizmettir. En bilinen kayıt defteri **Docker Hub**'dır. Kullanıcılar buradan genel imajları çekebilir (`docker pull`) veya kendi imajlarını yükleyebilir (`docker push`).
*   **Docker İmajları (Images):** Konteynerleri oluşturmak için kullanılan şablonlardır. Uygulamayı, kodunu, çalışma zamanını, sistem araçlarını, sistem kütüphanelerini ve bağımlılıkları içeren salt okunur katmanlı dosyalardır. Bir Dockerfile ile oluşturulurlar.
*   **Docker Konteynerleri (Containers):** Docker imajlarının çalıştırılabilir örnekleridir. İmajdan oluşturulur, izole edilmiş bir ortamda çalışır ve uygulamanın tüm bağımlılıklarını içerir.
*   **Dockerfile:** Docker imajlarını oluşturmak için kullanılan talimatları içeren bir metin dosyasıdır. Bir uygulamanın nasıl kurulacağını ve çalıştırılacağını adım adım açıklar.

### 2. Sık Kullanılan Docker Komutları

Docker CLI (Komut Satırı Arayüzü), Docker Engine ile etkileşim kurmak için kullanılır. İşte en sık kullanılan Docker komutlar ve açıklamaları:

*   `docker build [PATH]`: Bir `Dockerfile`'dan Docker imajı oluşturur. `[PATH]`, Dockerfile'ın bulunduğu dizindir.
    *   **Örnek:** `docker build -t my-streamlit-app .` (Mevcut dizindeki Dockerfile ile `my-streamlit-app` adında bir imaj oluşturur)

*   `docker run [OPTIONS] IMAGE [COMMAND] [ARG...]`: Yeni bir konteyner başlatır ve bir komut çalıştırır. Bir imajdan bir konteyner örneği oluşturur ve onu başlatır.
    *   **Örnek:** `docker run -p 8501:8501 my-streamlit-app` (8501 portunu host'a bağlayarak `my-streamlit-app` imajından bir konteyner çalıştırır)
    *   `-p`: Port haritalama (host_port:container_port)
    *   `-d`: Konteyneri arka planda çalıştırır (detached mode)
    *   `--name`: Konteynere özel bir isim verir

*   `docker pull NAME[:TAG]`: Bir imajı Docker Registry'den (örneğin Docker Hub) yerel makinenize çeker.
    *   **Örnek:** `docker pull ubuntu:latest`

*   `docker push NAME[:TAG]`: Yerel imajınızı bir Docker Registry'ye yükler.
    *   **Örnek:** `docker push myusername/my-streamlit-app:latest`

*   `docker images`: Yerel makinede bulunan tüm Docker imajlarını listeler.
    *   **Örnek:** `docker images`

*   `docker ps [OPTIONS]`: Çalışan konteynerleri listeler.
    *   **Örnek:** `docker ps`
    *   `-a`: Tüm konteynerleri (çalışan ve durdurulmuş) listeler.

*   `docker stop [CONTAINER_ID or CONTAINER_NAME]`: Çalışan bir konteyneri durdurur.
    *   **Örnek:** `docker stop my-streamlit-app-instance`

*   `docker rm [CONTAINER_ID or CONTAINER_NAME]`: Bir konteyneri siler (durdurulmuş olmasi gerekir).
    *   **Örnek:** `docker rm my-streamlit-app-instance`

*   `docker rmi [IMAGE_ID or IMAGE_NAME]`: Bir imajı siler.
    *   **Örnek:** `docker rmi my-streamlit-app`

*   `docker exec [OPTIONS] CONTAINER COMMAND [ARG...]`: Çalışan bir konteyner içinde bir komut çalıştırır.
    *   **Örnek:** `docker exec -it my-streamlit-app-instance bash` (Konteyner içinde interaktif bir bash oturumu başlatır)

*   `docker logs [CONTAINER_ID or CONTAINER_NAME]`: Bir konteynerin loglarını (çıktılarını) gösterir.
    *   **Örnek:** `docker logs my-streamlit-app-instance`

### 3. Streamlit Uygulamalarının Docker İçinde Çalışması

Bir Streamlit uygulamasını Docker içine almak (Dockerize etmek), uygulamanın tüm bağımlılıkları ve çalışma ortamı ile birlikte taşınabilir ve tutarlı bir paket halinde sunulmasını sağlar. İşte temel bir Streamlit uygulamasını Dockerize etmek için izlenen adımlar ve örnek bir `Dockerfile` içeriği:

**Streamlit Uygulaması İçin Dockerfile Adımları:**

1.  **Temel İmaj Seçimi (`FROM`):** Uygulamanızın ihtiyaç duyduğu temel işletim sistemini ve Python sürümünü içeren hafif bir temel imaj seçilir. Genellikle `python:3.x-slim-buster` veya `python:3.x-alpine` gibi imajlar tercih edilir.

    ```dockerfile
    FROM python:3.9-slim-buster
    ```

2.  **Çalışma Dizini Belirleme (`WORKDIR`):** Konteyner içinde uygulamanızın dosyalarının bulunacağı bir çalışma dizini tanımlanır.

    ```dockerfile
    WORKDIR /app
    ```

3.  **Gereksinimleri Kopyalama ve Yükleme (`COPY`, `RUN`):** `requirements.txt` dosyasını konteynere kopyalayıp, uygulamanın tüm Python bağımlılıklarını yüklersiniz. Bu, katmanlama (layering) avantajı sağlar; bağımlılıklar sık değişmiyorsa, bu katman tekrar oluşturulmaz ve build süresini hızlandırır.

    ```dockerfile
    COPY requirements.txt ./requirements.txt
    RUN pip install --no-cache-dir -r requirements.txt
    ```

4.  **Uygulama Kodunu Kopyalama (`COPY`):** Yerel makinenizdeki Streamlit uygulama dosyalarını (örneğin `app.py`) çalışma dizinine kopyalarsınız.

    ```dockerfile
    COPY . .
    ```

5.  **Ortam Değişkenleri ve Port Ayarları (`ENV`, `EXPOSE`):** Streamlit'in varsayılan olarak hangi portta çalışacağını belirtmek ve dışarıya açmak için kullanılır. Streamlit genellikle 8501 portunu kullanır.

    ```dockerfile
    ENV PORT 8501
    EXPOSE $PORT
    ```

6.  **Uygulamayı Çalıştırma Komutu (`CMD`):** Konteyner başlatıldığında çalıştırılacak varsayılan komut belirlenir. Streamlit uygulamasını başlatmak için `streamlit run` komutu kullanılır.

    ```dockerfile
    CMD ["streamlit", "run", "app.py", "--server.port", "$PORT", "--server.address", "0.0.0.0"]
    ```

**Örnek `Dockerfile`:**

```dockerfile
# 1. Temel imajı belirle
FROM python:3.9-slim-buster

# 2. Çalışma dizinini ayarla
WORKDIR /app

# 3. Bağımlılıkları kopyala ve yükle
COPY requirements.txt ./requirements.txt
RUN pip install --no-cache-dir -r requirements.txt

# 4. Uygulama kodunu kopyala
COPY . .

# 5. Portu belirle ve aç
ENV PORT 8501
EXPOSE $PORT

# 6. Streamlit uygulamasını başlat
CMD ["streamlit", "run", "app.py", "--server.port", "$PORT", "--server.address", "0.0.0.0"]
```

Bu `Dockerfile` ile bir imaj oluşturduktan (`docker build`) ve bu imajdan bir konteyner çalıştırdıktan (`docker run`) sonra, Streamlit uygulamanız izole edilmiş bir ortamda ve tutarlı bir şekilde çalışacaktır.

### 4. Konteynerleştirmenin Faydaları ve Streamlit Uygulamalarına Etkileri

Konteynerleştirme (Docker kullanımı), uygulama geliştirme ve dağıtım süreçlerine bir dizi önemli fayda sağlar. Bu faydalar, özellikle Streamlit gibi dinamik ve bağımlılık açısından zengin web uygulamaları için kritik öneme sahiptir.

**Konteynerleştirmenin Genel Faydaları:**

*   **Tutarlılık (Consistency):** Konteynerler, bir uygulamanın tüm çalışma zamanı ortamını (kod, bağımlılıklar, sistem kütüphaneleri, ayarlar) tek bir pakette birleştirir. Bu, geliştirme, test ve üretim ortamları arasında "benim makinemde çalışıyor" sorununu ortadan kaldırarak tutarlı bir çalışma sağlar.
*   **İzolasyon (Isolation):** Her konteyner birbirinden ve ana sistemden izole edilmiştir. Bu, farklı uygulamaların veya aynı uygulamanın farklı sürümlerinin birbirini etkilemeden aynı sunucuda güvenle çalışmasını sağlar. Bağımlılık çakışmaları riski azalır.
*   **Ölçeklenebilirlik (Scalability):** Konteynerler hafif ve hızlı başlangıç sürelerine sahiptir. Bu sayede, uygulamanın talebine göre hızla yeni konteyner örnekleri oluşturulabilir ve yük dengeleme ile ölçeklenebilir bir yapı kurulabilir. İhtiyaç kalmadığında da hızla sonlandırılabilirler.
*   **Taşınabilirlik (Portability):** Bir konteyner, herhangi bir Docker kurulu platformda (yerel bilgisayar, sanal makine, bulut ortamı) aynı şekilde çalışabilir. Bu, uygulamaların farklı altyapılar arasında kolayca taşınmasını ve dağıtılmasını sağlar.
*   **Verimlilik (Efficiency):** Sanal makinelere göre daha az kaynak tüketirler çünkü işletim sistemini paylaşırlar. Bu, daha fazla uygulamanın daha az donanım kaynağıyla çalışabilmesi anlamına gelir.
*   **Hızlı Dağıtım (Rapid Deployment):** Uygulamaların konteynerler halinde paketlenmesi, CI/CD (Sürekli Entegrasyon/Sürekli Dağıtım) süreçlerini basitleştirir ve uygulamaların daha hızlı bir şekilde canlıya alınmasını sağlar.

**Bu Faydaların Bir Streamlit Uygulaması İçin Geçerliliği:**

Streamlit uygulamaları genellikle Python tabanlıdır ve çeşitli Python kütüphanelerine (örneğin `streamlit`, `pandas`, `numpy`, `transformers`, `diffusers`, `torch`, `Pillow`) bağımlıdır. Bu bağımlılıklar, farklı geliştirme ortamlarında farklı sürümler veya çakışmalar nedeniyle sorunlara yol açabilir. Docker, bu sorunları çözmek için ideal bir araçtır:

1.  **Bağımlılık Yönetimi ve Tutarlılık:** Streamlit uygulamanızın gerektirdiği tüm Python kütüphaneleri ve diğer sistem bağımlılıkları `Dockerfile` içinde belirtilerek konteynere kurulur. Bu, uygulamanın her yerde (geliştirme dizüstü bilgisayarınızdan bulut sunucusuna kadar) tam olarak aynı ortamda çalışmasını garanti eder. `requirements.txt` dosyası ile bu süreç standartlaştırılır.

2.  **Kolay Dağıtım ve Paylaşım:** Geliştirdiğiniz Streamlit uygulamasını bir Docker imajına dönüştürdüğünüzde, bu imajı Docker Hub gibi bir kayıt defterine yükleyebilir ve ekibinizle veya genel kullanıcılarla kolayca paylaşabilirsiniz. Uygulamanızı çalıştırmak isteyen herkesin tek yapması gereken Docker imajını çekip bir konteyner olarak çalıştırmaktır; yerel bağımlılık kurulumlarıyla uğraşmalarına gerek kalmaz.

3.  **İzolasyon ve Güvenlik:** Eğer bir sunucuda birden fazla Streamlit uygulaması veya başka servisler çalıştırıyorsanız, her birini ayrı Docker konteynerlerinde çalıştırmak, uygulamaların birbirinin kaynaklarını etkilemesini veya güvenlik açıklarının yayılmasını önler. Örneğin, bir uygulamadaki bir kütüphane güncellemesi diğerini bozmaz.

4.  **Ölçeklenebilirlik:** Yüksek trafikli bir Streamlit uygulaması için, talebi karşılamak üzere birden fazla konteyner örneği hızla başlatılabilir. Bu, uygulamanızın performansını ve kullanıcı deneyimini artırır. Kubernetes gibi konteyner orkestrasyon araçlarıyla bu süreç otomatikleştirilebilir.

5.  **Ortam Basitliği:** Geliştiricilerin, Streamlit uygulamasını çalıştırmak için karmaşık yerel kurulumlar yapmak yerine sadece Docker'ı kurmaları yeterli olur. Bu, yeni ekip üyelerinin projeye adaptasyonunu hızlandırır ve geliştirme ortamlarının standartlaşmasını sağlar.

Sonuç olarak, Docker, Streamlit uygulamalarını daha güvenilir, taşınabilir ve ölçeklenebilir hale getirerek geliştirme ve dağıtım süreçlerini büyük ölçüde kolaylaştırır.

## Dockerfile Oluşturma

### Subtask:
Mevcut Streamlit uygulaması için bir `Dockerfile` oluşturun. Bu dosya, uygulamanın çalışması için gerekli bağımlılıkları (Python, pip paketleri gibi) ve uygulama kodunu içerecektir.


**Reasoning**:
The first step is to save the existing Streamlit application code into a file named `app.py` as instructed. This will make the application code available for the Dockerfile.



````python
app_code = '''
# Streamlit'i yükle (eğer kurulu değilse)
# !pip install -q streamlit
# !pip install -q transformers diffusers accelerate torch

import streamlit as st
import torch
from transformers import pipeline
from diffusers import StableDiffusionPipeline
from PIL import Image, ImageFilter
import os


# --- HAFTA 12: Sayfa Konfigürasyonu ve UI ---
st.set_page_config(page_title="Science & Art AI App", layout="wide")
st.title("🎨 Science and Art: Generative AI Suite")
st.sidebar.header("Haftalık Görev Menüsü")

# --- HAFTA 9: Mod Seçimi (Chat vs Art) ---
#  Chatbot ve Image Generator birleştirildi.
mode = st.sidebar.radio("Bir mod seçin:", ["Chat Mode (W5/9)", "Art Mode (W7/9/11)"])

# --- MODEL YÜKLEME (Caching ile hızlandırma) ---
@st.cache_resource
def load_chat_model():
    # Hafta 5: Ücretsiz Hugging Face modeli [cite: 2, 13]
    return pipeline("text2text-generation", model="google/flan-t5-base")

@st.cache_resource
def load_art_model():
    # Hafta 7: Stable Diffusion open-source model [cite: 2, 13]
    model_id = "runwayml/stable-diffusion-v1-5"
    pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
    if torch.cuda.is_available():
        pipe = pipe.to("cuda")
    return pipe

# --- CHAT MODE (Hafta 5 & 9) ---
if mode == "Chat Mode (W5/9)":
    st.header("💬 AI Chatbot")

    # HAFTA 10: Parametre Optimizasyonu
    # Sınavda sorulursa: temperature yaratıcılığı, top_p olasılık dağılımını belirler.
    with st.sidebar.expander("Prompt Ayarları (W10)"):
        temp = st.slider("Temperature (Yaratıcılık)", 0.1, 1.0, 0.7)
        max_len = st.slider("Maksimum Yanıt Uzunluğu", 20, 200, 50)

    user_query = st.text_input("Bir soru sorun:")
    if user_query:
        chat_model = load_chat_model()
        response = chat_model(user_query, max_length=max_len, temperature=temp)
        st.write("**Cevap:**", response[0]['generated_text'])

# --- ART MODE (Hafta 7, 9 & 11) ---
elif mode == "Art Mode (W7/9/11)":
    st.header("🖼️ Generative Art")

    prompt = st.text_input("Hayalinizdeki resmi tarif edin (İngilizce):", "A futuristic city in watercolor style")

    if st.button("Resim Oluştur"):
        with st.spinner("Sanat eseri hazırlanıyor..."):
            art_model = load_art_model()
            # Hafta 7: Resim üretme
            generated_image = art_model(prompt).images[0]

            # Resmi geçici olarak kaydet (Hafta 7 isterleri) [cite: 2, 9]
            if not os.path.exists("outputs"): os.makedirs("outputs")
            generated_image.save("outputs/last_generated.png")

            st.image(generated_image, caption="Orijinal Yapay Zeka Çıktısı")
            st.session_state['current_img'] = generated_image

    # --- HAFTA 11: Artistic Style Enhancement (Filtreler) ---
    #  Pillow/OpenCV kullanarak filtre ekleme.
    if 'current_img' in st.session_state:
        st.divider()
        st.subheader("Sanatsal Filtreler (W11)")
        filter_type = st.selectbox("Bir filtre seçin:", ["Orijinal", "Siyah-Beyaz", "Bulanık (Blur)", "Kenar Belirleme"])

        processed_img = st.session_state['current_img'].copy()

        if filter_type == "Siyah-Beyaz":
            processed_img = processed_img.convert("L")
        elif filter_type == "Bulanık (Blur)":
            processed_img = processed_img.filter(ImageFilter.BLUR)
        elif filter_type == "Kenar Belirleme":
            processed_img = processed_img.filter(ImageFilter.FIND_EDGES)

        st.image(processed_img, caption=f"Uygulanan Filtre: {filter_type}")
'''

with open('app.py', 'w') as f:
    f.write(app_code)

print("Streamlit application code saved to app.py")
````

**Reasoning**:
The next step is to create a `requirements.txt` file that lists all the Python dependencies required by the Streamlit application, as instructed in the subtask. This file will be used by Docker to install the necessary packages.



````python
requirements = [
    "streamlit",
    "transformers",
    "diffusers",
    "accelerate",
    "torch",
    "Pillow"
]

with open('requirements.txt', 'w') as f:
    for req in requirements:
        f.write(req + '\n')

print("requirements.txt created successfully.")
````

**Reasoning**:
The final step for creating the Dockerfile is to write its content based on the specified instructions, including the base image, working directory, dependency installation, code copying, port configuration, and the command to run the Streamlit application.



````python
dockerfile_content = '''
# 1. Temel imajı belirle
FROM python:3.9-slim-buster

# 2. Çalışma dizinini ayarla
WORKDIR /app

# 3. Bağımlılıkları kopyala ve yükle
COPY requirements.txt ./requirements.txt
RUN pip install --no-cache-dir -r requirements.txt

# 4. Uygulama kodunu kopyala
COPY app.py .

# Kütüphane klasörlerini kopyala (eğer varsa, Stable Diffusion için)
# COPY models/ models/
# COPY outputs/ outputs/

# 5. Portu belirle ve aç
ENV PORT 8501
EXPOSE $PORT

# 6. Streamlit uygulamasını başlat
CMD ["streamlit", "run", "app.py", "--server.port", "$PORT", "--server.address", "0.0.0.0"]
'''

with open('Dockerfile', 'w') as f:
    f.write(dockerfile_content)

print("Dockerfile created successfully.")
````

### Dockerfile'ı ve Uygulama Dosyalarını GitHub'a Yükleme

**Adım 1: Proje Dizinini Hazırlama**

Öncelikle, oluşturduğunuz `Dockerfile`, `app.py` ve `requirements.txt` dosyalarının aynı dizinde olduğundan emin olun.

**Adım 2: Git Deposunu Başlatma**

Bu dizine gidin ve bir Git deposu başlatın. (Eğer dizinde zaten bir Git deposu varsa bu adımı atlayabilirsiniz.)

```bash
git init
```

**Adım 3: Dosyaları Git Deposuna Ekleme**

Şimdi `Dockerfile`, `app.py` ve `requirements.txt` dosyalarını Git'in takip etmesi için ekleyin:

```bash
git add Dockerfile app.py requirements.txt
```

**Adım 4: Değişiklikleri Kaydetme (Commit)**

Eklediğiniz dosyaları bir commit mesajıyla kaydedin:

```bash
git commit -m "Add Dockerfile and Streamlit app files"
```

**Adım 5: GitHub'da Yeni Bir Depo Oluşturma**

GitHub hesabınıza gidin ve *yeni, boş bir depo* oluşturun. Depo oluştururken **README, .gitignore veya lisans dosyası eklemeyin**, aksi takdirde push işlemi sırasında çakışmalar yaşanabilir.

**Adım 6: Yerel Depoyu Uzak Depoya Bağlama**

GitHub'da deponuzu oluşturduktan sonra size verilen komutları kullanarak yerel deponuzu bu uzak depoya bağlayın. Genellikle şu formatta olacaktır:

```bash
git remote add origin https://github.com/YOUR_USERNAME/YOUR_REPOSITORY_NAME.git
```

**Adım 7: Dosyaları GitHub'a Yükleme**

Son olarak, yerel değişikliklerinizi GitHub'daki uzak depoya yükleyin:

```bash
git push -u origin main
# Veya eğer varsayılan dal adınız 'master' ise:
git push -u origin master
```

Bu adımları tamamladığınızda, `Dockerfile`, `app.py` ve `requirements.txt` dosyalarınız GitHub deponuza başarıyla yüklenmiş olacaktır.
